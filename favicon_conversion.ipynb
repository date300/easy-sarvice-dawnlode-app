```python

This notebook loads the existing PNG logo, prepares standard favicon sizes, and saves a single ICO file at `public/favicon.ico`.
</VSCode.Cell>
<VSCode.Cell id="2" language="python">from PIL import Image
from pathlib import Path

source_path = Path("public/EasyService.png")
output_path = Path("public/favicon.ico")

print(f"Source path: {source_path.resolve()}")
print(f"Output path: {output_path.resolve()}")
print(f"Source exists: {source_path.exists()}")
</VSCode.Cell>
<VSCode.Cell id="3" language="markdown">## Load source PNG image

Read the source image and inspect its mode and size to ensure it can be converted correctly.
</VSCode.Cell>
<VSCode.Cell id="4" language="python">image = Image.open(source_path)
print(f"Image format: {image.format}")
print(f"Image mode: {image.mode}")
print(f"Image size: {image.size}")
image
</VSCode.Cell>
<VSCode.Cell id="5" language="markdown">## Prepare favicon icon sizes

Resize the image to multiple icon sizes commonly used for favicons.
</VSCode.Cell>
<VSCode.Cell id="6" language="python">sizes = [(16, 16), (32, 32), (48, 48)]
resized_images = [image.resize(size, Image.LANCZOS) for size in sizes]
for size, img in zip(sizes, resized_images):
    print(f"Prepared size: {size} -> {img.size}")
</VSCode.Cell>
<VSCode.Cell id="7" language="markdown">## Save ICO file as public/favicon.ico

Save the resized images into a single ICO file.
</VSCode.Cell>
<VSCode.Cell id="8" language="python">output_path.parent.mkdir(parents=True, exist_ok=True)
resized_images[0].save(output_path, format="ICO", sizes=sizes)
print(f"Saved favicon.ico: {output_path.exists()}")
</VSCode.Cell>
<VSCode.Cell id="9" language="markdown">## Verify favicon output

Confirm that the ICO file exists and inspect its size.
</VSCode.Cell>
<VSCode.Cell id="10" language="python">print(output_path.stat())
print("favicon.ico created successfully.")
```
